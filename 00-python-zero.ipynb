```python
def compter_voyelles(mot):
  nombre = 0
  for i in mot:
    if i == 'a' or i == 'e' or i =='e' or i == 'o' or i == 'u' or i == 'A' or i == 'E' or i =='I' or i == 'O' or i == 'U':
      nombre = nombre + 1
  return nombre

print(compter_voyelles("mouche"))
```
